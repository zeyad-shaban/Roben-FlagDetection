```python
import torch
import torch.nn as nn
import torch.optim as optim
from jax.example_libraries.optimizers import optimizer
from numpy import dtype
from torchvision import transforms, models
from torch.utils.data import Dataset, DataLoader
from torchsummary import summary

from glob import glob
import os

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageEnhance

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
### HYPER PARAMS ###
batch_size = 128
```

```python
class BackgroundWithFlag(Dataset):
    def __init__(self, bg_path: str, flag_path: str):
        self.p_flag = 0.4
        self.p_augment_background = 0.7
        self.p_augment_flag = 0.8
        
        self.flag_min_size = 7
        self.flag_max_size = 41
        self.flag_max_shear = 0.4 

        self.backgrounds = glob(os.path.join(bg_path, '*.jpg'))
        self.flags = glob(os.path.join(flag_path, '*.png'))

        self.all_augmentation = transforms.Compose([
            transforms.ToTensor(),
            transforms.GaussianBlur(3),
            transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
        ])

        self.background_augmentation = transforms.Compose([
            transforms.Resize(256),
            transforms.CenterCrop(224),

            transforms.RandomHorizontalFlip(),
            transforms.RandomVerticalFlip(),
            transforms.RandomAffine(degrees=0, translate=(0.05, 0.05), scale=(1.1, 1.2), shear=5),
            transforms.ColorJitter(brightness=(0.3, 1.4), contrast=(0.4, 1.5), saturation=(0.4, 1.6), hue=(-0.1, 0.1)),
        ])

    def __len__(self):
        return len(self.backgrounds)

    # todo use collate function for augmentation

    def __getitem__(self, index):
        background = Image.open(self.backgrounds[index]).convert("RGB")
        if np.random.random() <= self.p_augment_background:
            background = self.background_augmentation(background)
        else:
            background = self.background_augmentation[0:5](background)

        flag_center = (-1, -1)
        flag_exists = 0
        composite_image = background

        if np.random.random() <= self.p_flag:
            flag = Image.open(np.random.choice(self.flags)).convert("RGB")  # Flag with transparency
            
            if np.random.random() <= self.p_augment_flag:
                flag = self.augment_flag(flag)
            else:
                flag = flag.resize((np.random.randint(self.flag_min_size, self.flag_max_size), np.random.randint(self.flag_min_size, self.flag_max_size))).convert('RGBA')

            composite_image, flag_center = self.paste_flag(background, flag)
            flag_exists = 1

        composite_image = self.all_augmentation(composite_image)

        return (
            composite_image.to(device),
            torch.tensor(flag_center, dtype=torch.float32, device=device),
            torch.tensor(flag_exists, dtype=torch.float32, device=device)
        )

    def augment_flag(self, flag: Image.Image):
        # Photometric Augmentations
        if np.random.random() >= 0.5:
            flag = flag.transpose(Image.FLIP_LEFT_RIGHT)
        
        if np.random.random() >= 0.5:
             flag = flag.transpose(Image.FLIP_TOP_BOTTOM)
            
        enhancer = ImageEnhance.Brightness(flag)
        flag = enhancer.enhance(np.random.uniform(0.3, 1.5))
    
        # Contrast adjustment
        enhancer = ImageEnhance.Contrast(flag)
        flag = enhancer.enhance(np.random.uniform(0.3, 2))
    
        # Saturation adjustment
        enhancer = ImageEnhance.Color(flag)
        flag = enhancer.enhance(np.random.uniform(0.4, 1.5))
    
        # Hue adjustment (we simulate this by shifting colors in HSV space)
        hue_shift = np.random.uniform(-0.5, 0.5)
        flag = flag.convert('HSV')
        channels = list(flag.split())
        channels[0] = channels[0].point(lambda h: (h + hue_shift * 255) % 255)
        flag = Image.merge('HSV', channels).convert('RGBA')
        
        # Affine Augmentation
        flag = flag.resize((np.random.randint(self.flag_min_size, self.flag_max_size), np.random.randint(self.flag_min_size, self.flag_max_size)))
        
        flag = flag.rotate(np.random.randint(0, 360), expand=True)
        
        # Shear
        width, height = flag.size
        shear_factor = np.random.uniform(-self.flag_max_shear, self.flag_max_shear)  # Random shear factor
        matrix = (1, shear_factor, 0, shear_factor, 1, 0)
        flag = flag.transform((width, height), Image.AFFINE, matrix, resample=Image.BICUBIC)
        
        return flag
    
    @staticmethod
    def paste_flag(background, flag):
        bg_w, bg_h = background.size
        flag_w, flag_h = flag.size

        max_x = bg_w - flag_w
        max_y = bg_h - flag_h
        x_pos = np.random.randint(0, max_x)  # max_x + flag_w to have the last half only
        y_pos = np.random.randint(0, max_y)  # max_y + flag_h to have the last half only

        background.paste(flag, (x_pos, y_pos), flag)

        flag_center = (x_pos + flag_w // 2, y_pos + flag_h // 2)
        return background, flag_center


dataset = BackgroundWithFlag('./data/desert', './data/flags')
dataset[0]
dataloader = DataLoader(dataset, batch_size, True)
```

```python
model = models.googlenet(weights=models.GoogLeNet_Weights.DEFAULT)

for param in model.parameters():
    param.requires_grad = False
```

```python
class FCLayer(nn.Module):
    def __init__(self):
        super().__init__()

        self.fc_base = nn.Sequential(
            nn.Dropout(0.2),
            nn.Linear(1024, 256),
            nn.BatchNorm1d(256),
            nn.Mish(),
        )

        self.pos_layer = nn.Sequential(
            nn.Dropout(0.2),
            nn.Linear(256, 64),
            nn.BatchNorm1d(64),
            nn.Mish(),

            nn.Dropout(0.2),
            nn.Linear(64, 2),  # x, y
            nn.ReLU(),
        )

        self.exists_layer = nn.Sequential(
            nn.Dropout(0.2),
            nn.Linear(256, 64),
            nn.BatchNorm1d(64),
            nn.Mish(),

            nn.Dropout(0.2),
            nn.Linear(64, 1),
            nn.Sigmoid(),
        )

    def forward(self, x):
        x = self.fc_base(x)
        pos = self.pos_layer(x)
        exists = self.exists_layer(x)

        return pos, exists


model.fc = FCLayer().to(device)
```

```python
opt = optim.Adam(model.fc.parameters(), lr=1e-1)
criterion_exists = nn.BCELoss()
criterion_pos = nn.MSELoss()
lr_schedule = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer=opt,
    factor=0.5,
    patience=10,
    threshold=1e-3,
    threshold_mode='rel',
    min_lr=1e-7,
)
```

```python
epochs = 5
losses_pos = []
losses_exists = []
losses_total = []

for epoch in range(epochs):
    for img, pos, exists in dataloader:
        _pos, _exists = model(img)
        loss_pos = criterion_pos(_pos, pos) * exists
        loss_exists = criterion_exists(_exists.squeeze(-1), exists) * 3

        losses_pos.append(loss_pos.item())
        losses_exists.append(loss_exists.item())

        loss_total = loss_pos + loss_exists

        losses_total.append(loss_total.item())

        opt.zero_grad()
        loss_total.backward()
        opt.step()

        lr_schedule.step(loss_total.item())
        print('done with a batch')
    print(
        f'Epoch: {epoch}, Total Loss: {losses_total[-1]}, Loss pos: {losses_pos[-1]}, Loss exists: {losses_exists[-1]}, LR: {opt.param_groups[0]["lr"]}')
```

```python
plt.figure(figsize=(6, 6))
plt.subplot(221)
plt.plot(range(len(losses_pos)), losses_pos)
plt.title('Loss Position')
plt.xlabel('Epoch')
plt.ylabel('Loss')

plt.subplot(222)
plt.plot(range(len(losses_exists)), losses_exists)
plt.title('Loss Exists')
plt.xlabel('Epoch')
plt.ylabel('Loss Exists')

plt.subplot(223)
plt.plot(range(len(losses_total)), losses_total)
plt.title('Loss Total')
plt.xlabel('Epoch')
plt.ylabel('Loss Total')

plt.tight_layout()
```
