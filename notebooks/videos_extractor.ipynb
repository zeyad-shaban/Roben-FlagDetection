```python
import cv2
import os
```

```python
def extract_frames(video_path, output_dir, fps=2, id=0):
    """
    Extracts frames from a video at the specified FPS.
    :param video_path: Path to the input video.
    :param output_dir: Directory to save the extracted frames.
    :param fps: Number of frames per second to extract.
    """
    # Create output directory if it doesn't exist
    os.makedirs(output_dir, exist_ok=True)

    # Load video
    video = cv2.VideoCapture(video_path)
    video_fps = video.get(cv2.CAP_PROP_FPS)
    frame_interval = int(video_fps / fps)

    frame_count = 0
    saved_frame_count = 0

    while True:
        success, frame = video.read()
        if not success:
            break

        # Save frame at the specified interval
        if frame_count % frame_interval == 0:
            frame_name = f"frame_{id}_{saved_frame_count:04d}.jpg"
            frame_path = os.path.join(output_dir, frame_name)
            
            h, w, _ = frame.shape
            frame = cv2.resize(frame, (w // 2, h // 2))
            cv2.imwrite(frame_path, frame)
            saved_frame_count += 1

        frame_count += 1

    video.release()
    print(f"Extracted {saved_frame_count} frames to {output_dir}")

# Example usage
video_path = "../data/noury/videos/flag_img_1.MP4"  # Path to your video file
output_dir = "../data/noury/images"  # Directory to save frames
extract_frames(video_path, output_dir, fps=1, id=1)
```
