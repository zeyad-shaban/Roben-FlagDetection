```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import cv2
```

```python
size = 500
in_path = "/home/zeyadcode/Videos/RobEn/GX011072.MP4"
out_path = f"./data/cam_vid_{size}_{size}.mp4"

cap = cv2.VideoCapture(in_path)
fps = cap.get(cv2.CAP_PROP_FPS)

fourcc = cv2.VideoWriter_fourcc(*"mp4v")  # or 'XVID'
out = cv2.VideoWriter(out_path, fourcc, fps, (size, size))

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Resize the frame
    resized_frame = cv2.resize(frame, (size, size))

    # Write the frame to output
    out.write(resized_frame)

cap.release()
out.release()
```
