```python
import cv2
import os
```

```python
# Configuration
video_path = "./data/cam_vid_500_500.mp4"    # Replace with your video path
output_dir = "./data/raw_frames"             # Directory to save frames

os.makedirs(output_dir, exist_ok=True)

cap = cv2.VideoCapture(video_path)
if not cap.isOpened():
    raise Exception("Error: Could not open video")

frame_id = 0
while True:
    success, frame = cap.read()
    
    if not success:
        break
    
    frame_name = f"frame_{frame_id:06d}.jpg"
    output_path = os.path.join(output_dir, frame_name)
    cv2.imwrite(output_path, frame)
    
    frame_id += 1

cap.release()
print(f"Saved {frame_id} frames to '{output_dir}'")
```
